1. Array create (10,20,30,40,50)

```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
print(arr)
```

2. arange() 1 to 10

```python
arr = np.arange(1, 11)
print(arr)
```

3. 5 zeros array


```python
arr = np.zeros(5)
print(arr)
```

4. 5 ones array

```python
arr = np.ones(5)
print(arr)
```

5. Even numbers 2 to 20

```python
arr = np.arange(2, 21, 2)
print(arr)
```

6. Number of elements

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr.size)
```

7. Shape and dimensions

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print("Shape:", arr.shape)
print("Dimensions:", arr.ndim)
```

8. 3x3 array from 1 to 9

```python
arr = np.arange(1, 10).reshape(3, 3)
print(arr)
```

9. Max and min values

```python
arr = np.array([15, 8, 25, 10, 30])
print("Max:", arr.max())
print("Min:", arr.min())
```

10. Sum and average

```python
arr = np.array([10, 20, 30, 40, 50])
print("Sum:", arr.sum())
print("Average:", arr.mean())
```

11. Index of maximum value

```python
arr = np.array([15, 8, 25, 10, 30])
print(arr.argmax())
```

12. First, third, last elements

```python
arr = np.array([100, 200, 300, 400, 500])
print(arr[0], arr[2], arr[-1])
```

13. Reverse the array

```python
arr = np.array([1, 2, 3, 4, 5])
print(arr[::-1])
```

14. First three elements (slicing)

```python
arr = np.array([1, 2, 3, 4, 5])
print(arr[:3])
```

15. 2x4 array, rows and columns

```python
arr = np.arange(1, 9).reshape(2, 4)
print(arr)
print("Rows:", arr.shape[0])
print("Columns:", arr.shape[1])
```

16. 1 to 20, extract even numbers

```python
arr = np.arange(1, 21)
even = arr[arr % 2 == 0]
print(even)
```

17. Replace values > 50 with 0

```python
arr = np.array([10, 60, 30, 80, 40, 90])
arr[arr > 50] = 0
print(arr)
```

18. Values greater than 25

```python
arr = np.array([10, 30, 15, 40, 20, 50])
print(arr[arr > 25])
```

19. Addition, subtraction, multiplication, division

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])
print("Add:", a + b)
print("Sub:", a - b)
print("Mul:", a * b)
print("Div:", a / b)
```

20. Element-wise square

```python
arr = np.array([2, 4, 6, 8, 10])
print(arr ** 2)
```

21. Square root of every value

```python
arr = np.array([4, 9, 16, 25])
print(np.sqrt(arr))
```

22. Mean, median, std dev, variance

```python
arr = np.array([10, 20, 30, 40, 50])
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Std Dev:", np.std(arr))
print("Variance:", np.var(arr))
```

24. 3x3 matrix, sum of each column

26. Min value from each column

27. 1D array to 2x3 using reshape()

```python
arr = np.array([1, 2, 3, 4, 5, 6])
print(arr.reshape(2, 3))
```

28. Flatten 2D array to 1D

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr.flatten())
```

29. 2x2 matrix multiplication

```python
m1 = np.array([[1, 2], [3, 4]])
m2 = np.array([[5, 6], [7, 8]])
print(np.dot(m1, m2))
```

30. Marks analysis

```python
marks = np.array([45, 78, 92, 35, 88, 67, 30, 95])
print("Above 60:", marks[marks > 60])
print("Highest:", marks.max())
print("Lowest:", marks.min())
print("Average:", marks.mean())
print("Count above 60:", (marks > 60).sum())
```
