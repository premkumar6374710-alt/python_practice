5. Library Book

```python
books = ["Python", "Java", "C", "HTML"]

choice = input("Do you want a book? (yes/no): ")

if choice == "yes":
    card = input("Do you have a library card? (yes/no): ")

    if card == "yes":
        print("Available Books:", books)
        book = input("Enter book name: ")

        if book in books:
            books.remove(book)
            print(book, "issued successfully")
            print("Remaining Books:", books)
        else:
            print("Book not available")
    else:
        print("Library card required")
else:
    print("Thank You")
```

6. Factorial

```python
num = int(input("Enter a number: "))
fact = 1

for i in range(1, num + 1):
    fact = fact * i

print("Factorial =", fact)
```

7. Armstrong Number

```python
num = int(input("Enter a number: "))
temp = num
total = 0

while temp > 0:
    digit = temp % 10
    total = total + digit ** 3
    temp = temp // 10

if total == num:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

9. Sum of Digits

```python
num = int(input("Enter a number: "))
total = 0

while num > 0:
    digit = num % 10
    total = total + digit
    num = num // 10

print("Sum of digits =", total)
```

13. Count Vowels

```python
def count_vowels(text):
    count = 0

    for ch in text.lower():
        if ch in "aeiou":
            count += 1

    return count

text = input("Enter a string: ")

print("Vowels =", count_vowels(text))
```

14. Prime Number Function

```python
def is_prime(n):

    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

num = int(input("Enter a number: "))

print(is_prime(num))
```

15. Simple Calculator

```python
def calculator(a, b, op):

    if op == "+":
        return a + b

    elif op == "-":
        return a - b

    elif op == "*":
        return a * b

    elif op == "/":
        return a / b

    else:
        return "Invalid Operator"

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
operator = input("Enter operator (+,-,*,/): ")

print("Result =", calculator(num1, num2, operator))
```
