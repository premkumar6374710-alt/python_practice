25. 10 numbers - even, odd, max, min

```python
nums = list(map(int, input("Enter 10 numbers: ").split()))
even = [n for n in nums if n%2==0]
odd = [n for n in nums if n%2!=0]
print("Even:", even)
print("Odd:", odd)
print("Max:", max(nums))
print("Min:", min(nums))
```
