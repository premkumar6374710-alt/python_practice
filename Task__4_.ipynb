```python
# 1. Create a Pandas Series
import pandas as pd
import numpy as np

students = pd.Series(['Arun', 'Bala', 'Chitra', 'Divya', 'Eswar'])
print(students)
```

```python
# 2. Create DataFrame
data = {
    'Name': ['Arun', 'Bala', 'Chitra', 'Divya', 'Eswar', 'Farida', 'Gowtham', 'Hari', 'Indhu', 'Jeeva'],
    'Age': [21, 22, 20, 23, 21, 24, 22, np.nan, 20, 23],
    'Gender': ['M', 'M', 'F', 'F', 'M', 'F', 'M', 'M', 'F', 'M'],
    'Marks': [85, 45, 92, 67, 38, 78, np.nan, 55, 88, 70],
    'Department': ['CSE', 'ECE', 'CSE', 'IT', 'ECE', 'CSE', 'IT', 'ECE', 'CSE', 'IT'],
    'Join_Date': ['2023-01-15', '2023-02-20', '2023-01-10', '2023-03-05',
                  '2023-02-18', '2023-01-25', '2023-03-12', '2023-02-01',
                  '2023-01-30', '2023-03-20']
}
df = pd.DataFrame(data)
print(df)
```

```python
# 17. Age > 20
print(df[df['Age'] > 20])
```

```python
# 18. CSE department
print(df[df['Department'] == 'CSE'])
```

```python
# 19. Marks between 60 and 80
print(df[(df['Marks'] >= 60) & (df['Marks'] <= 80)])
```

```python
# 20. Two conditions with &
print(df[(df['Marks'] > 60) & (df['Gender'] == 'F')])
```

```python
# 21. Null values count
print(df.isnull().sum())
```

```python
# 22. Fill numerical missing with mean
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
df['Age'] = df['Age'].fillna(df['Age'].mean())
print(df)
```

```python
# 23. Fill categorical missing with mode
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])
print(df)
```

```python
# 24. Duplicate rows
print(df.duplicated().sum())
df = df.drop_duplicates()
print(df)
```

```python
# 25. Rename columns
df = df.rename(columns={'Marks': 'Total_Marks'})
print(df.columns)
```

```python
# 26. Result column
df['Result'] = df['Total_Marks'].apply(lambda x: 'Pass' if x >= 40 else 'Fail')
print(df)
```

```python
# 27. Age_Group column
df['Age_Group'] = df['Age'].apply(lambda x: 'Below 21' if x < 21 else ('21-23' if x <= 23 else 'Above 23'))
print(df)
```

```python
# 28. Avg marks department-wise
print(df.groupby('Department')['Total_Marks'].mean())
```

```python
# 29. Max marks per department
print(df.groupby('Department')['Total_Marks'].max())
```

```python
# 30. Department & Gender combo count
print(df.groupby(['Department', 'Gender']).size())
```

```python
# 31. Highest marks student in each department
print(df.loc[df.groupby('Department')['Total_Marks'].idxmax()])
```

```python
# 32. Top 3 students
print(df.sort_values('Total_Marks', ascending=False).head(3))
```

```python
# 33. Avg marks male vs female
print(df.groupby('Gender')['Total_Marks'].mean())
```

```python
# 34. Departments where avg marks > 70
avg = df.groupby('Department')['Total_Marks'].mean()
print(avg[avg > 70])
```

```python
# 35. Grade column
def grade(m):
    if m >= 90: return 'A'
    elif m >= 75: return 'B'
    elif m >= 60: return 'C'
    elif m >= 40: return 'D'
    else: return 'F'

df['Grade'] = df['Total_Marks'].apply(grade)
print(df)
```

```python
# 36. Pass/Fail column
df['Pass_Fail'] = df['Total_Marks'].apply(lambda x: 'Pass' if x >= 40 else 'Fail')
print(df)
```

```python
# 37. Extract year, month, day
df['Join_Date'] = pd.to_datetime(df['Join_Date'])
df['Year'] = df['Join_Date'].dt.year
df['Month'] = df['Join_Date'].dt.month
df['Day'] = df['Join_Date'].dt.day
print(df)
```

```python
# 38. Groupby avg, max, min together
print(df.groupby('Department')['Total_Marks'].agg(['mean', 'max', 'min']))
```

```python
# 39. Pivot table - avg marks by department and gender
pivot = pd.pivot_table(df, values='Total_Marks', index='Department', columns='Gender', aggfunc='mean')
print(pivot)
```

```python
# 40. Complete data cleaning and save as CSV
df['Age'] = df['Age'].astype(int)
df['Gender'] = df['Gender'].str.strip().str.upper()
df['Department'] = df['Department'].str.strip().str.upper()
df.to_csv('cleaned_students.csv', index=False)
print(df)
```
