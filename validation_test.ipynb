```python
num = 7
if num % 2 == 0:
    print(num, "is Even")
else:
    print(num, "is Odd")
```

5. Largest of three numbers if-elif-else

```python
a, b, c = 12, 45, 30
if a >= b and a >= c:
    print("Largest:", a)
elif b >= a and b >= c:
    print("Largest:", b)
else:
    print("Largest:", c)
```

6. Leap year check

```python
year = 2024
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(year, "is a Leap Year")
else:
    print(year, "is not a Leap Year")
```

7. Grade calculation

```python
marks = 82
if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 40:
    grade = "D"
else:
    grade = "Fail"
print("Grade:", grade)
```
