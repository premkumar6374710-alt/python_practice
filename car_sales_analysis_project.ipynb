```python
# Load dataset
import pandas as pd
import numpy as np

df = pd.read_csv('Car_sales.csv')
print("Dataset Loaded!")
print("Shape:", df.shape)
df.head()
```

```python
#Basic Data Cleaning
df.columns = df.columns.str.strip()

for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()
```

```python
#Handle Missing Values
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].mean())

for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values after handling:\n", df.isnull().sum())
```

```python
#Remove Duplicate Values
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

```python
#Convert Data Types if Required
# Example: df['Date'] = pd.to_datetime(df['Date'])
# Example: df['Price'] = pd.to_numeric(df['Price'], errors='coerce')
print(df.dtypes)
```

```python
#Numerical Column Analysis - Mean, Median, Min, Max
num_cols = df.select_dtypes(include=np.number).columns

print("Mean:\n", df[num_cols].mean())
print("\nMedian:\n", df[num_cols].median())
print("\nMin:\n", df[num_cols].min())
print("\nMax:\n", df[num_cols].max())
```

```python
#Numerical Column Analysis - Standard Deviation
print("Standard Deviation:\n", df[num_cols].std())
```

```python
#Categorical Column Analysis - Value Counts
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    print(f"\nValue Counts for {col}:")
    print(df[col].value_counts())
```

```python
#Categorical Column Analysis - Frequency (Percentage)
for col in cat_cols:
    print(f"\nFrequency (%) for {col}:")
    print(df[col].value_counts(normalize=True) * 100)
```

```python
# Step 5 - Important Relationships Find

# Create correlation matrix
correlation = df.corr(numeric_only=True)

# Copy correlation matrix
corr_pairs = correlation.copy()

# Remove self-correlation
import numpy as np
np.fill_diagonal(corr_pairs.values, np.nan)

# Find highest positive correlation
highest_positive = corr_pairs.stack().idxmax()
positive_value = corr_pairs.stack().max()

# Find highest negative correlation
highest_negative = corr_pairs.stack().idxmin()
negative_value = corr_pairs.stack().min()

# Display results
print("Highest Positive Correlation:")
print(highest_positive, "=", round(positive_value, 2))

print("\nHighest Negative Correlation:")
print(highest_negative, "=", round(negative_value, 2))
```

```python
# Step 5 & Step 6 - Find Important Relationships and Interpretation

# Create correlation matrix
corr_matrix = df.corr(numeric_only=True)

# Remove self-correlation (diagonal values = 1)
corr_no_diag = corr_matrix.copy()

import numpy as np
np.fill_diagonal(corr_no_diag.values, np.nan)

# Find highest positive correlation
highest_positive = corr_no_diag.stack().idxmax()
positive_value = corr_no_diag.stack().max()

# Find highest negative correlation
highest_negative = corr_no_diag.stack().idxmin()
negative_value = corr_no_diag.stack().min()

# Step 6 - Interpretation
print(f"Highest Positive Correlation is between {highest_positive[0]} and {highest_positive[1]} with value {positive_value:.2f}")
print("→ Indha rendu variables um ore direction la move aagum (oru increase aana, matradhu increase aagum)")

print(f"\nHighest Negative Correlation is between {highest_negative[0]} and {highest_negative[1]} with value {negative_value:.2f}")
print("→ Indha rendu variables um opposite direction la move aagum (oru increase aana, matradhu decrease aagum)")
```

```python
#outlier analysis - boxplot and IQR method to detect outliers
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("Car_sales.csv")
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols[:4]:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(f"{col}: {len(outliers)} outliers found")

    plt.figure(figsize=(5,3))
    plt.boxplot(df[col].dropna())
    plt.title(f"Outlier Check - {col}")
    plt.tight_layout()
    plt.show()
```

```python
#trend analysis - trend line if date/year column exists, else index-based trend
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("Car_sales.csv")
numeric_cols = df.select_dtypes(include=np.number).columns
num_col = numeric_cols[0]

date_like_cols = [c for c in df.columns if 'year' in c.lower() or 'date' in c.lower()]

if date_like_cols:
    trend_col = date_like_cols[0]
    trend = df.groupby(trend_col)[num_col].mean()
else:
    trend = df[num_col]

plt.figure(figsize=(8,5))
trend.plot(kind='line', marker='o', color='teal')
plt.title(f"Trend Analysis - {num_col}")
plt.ylabel(num_col)
plt.tight_layout()
plt.show()
```
